```python

%load_ext autoreload
%reload_ext autoreload
%autoreload 2

import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parents[1]))  # workaround to resolve my helper module above

import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import random

from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import KFold, cross_val_score, cross_validate, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer, mean_squared_error
import numpy as np
import pickle

from modules.plotting import *

DATA_PATH = "./data/"
MODELS_PATH = "./models/"
SUBMISSIONS_PATH = "./submissions/"
RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)
```

```python
training_data = pl.read_csv(DATA_PATH + "pca_one_hot_df.csv")
test_data = pl.read_csv(DATA_PATH + "test.csv")
```

```python
X = training_data.drop("quality_score")
y = training_data["quality_score"]
```

```python
baseline_lr = LinearRegression()

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)

scoring = {
    "r2": "r2",
    "rmse": "neg_root_mean_squared_error"
}

cv_baseline_lr_res = cross_validate(baseline_lr, X, y, cv=kfold, scoring=scoring, n_jobs=-1, return_estimator=True)

rmse_scores_baseline = -cv_baseline_lr_res["test_rmse"]
r2_scores_baseline = cv_baseline_lr_res["test_r2"]

print("Average R²:", r2_scores_baseline.mean())
print("Average RMSE:", rmse_scores_baseline.mean())
```

```python
baseline_lasso = Lasso()

cv_baseline_lasso_res = cross_validate(baseline_lasso, X, y, cv=kfold, scoring=scoring, n_jobs=-1, return_estimator=True)

rmse_scores_lasso_baseline = -cv_baseline_lasso_res["test_rmse"]
r2_scores_lasso_baseline = cv_baseline_lasso_res["test_r2"]

print("Average R²:", r2_scores_lasso_baseline.mean())
print("Average RMSE:", rmse_scores_lasso_baseline.mean())
```
